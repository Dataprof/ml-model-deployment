```python
import pandas as pd
import numpy as np
import tensorflow as tf
dataset = pd.read_csv('https://raw.githubusercontent.com/Dataprof/ml-model-deployment/main/storepurchasedata_large.csv')
X=dataset.iloc[:,:-1].values
y=dataset.iloc[:,-1].values

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=0)

from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)

model=tf.keras.models.Sequential([
tf.keras.layers.Dense(10,activation='relu'),
tf.keras.layers.Dense(10,activation='relu'),
tf.keras.layers.Dense(2,activation='softmax') 
])
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
model.fit(X_train,y_train,epochs=50)
loss,accuracy =  model.evaluate(X_test,y_test)
model.summary()
predictions=model.predict(sc.transform(np.array([[42,50000]])))[:,1]
predictions
model.save('customer_behavior_model.keras')
from tensorflow.keras.models import load_model
new_model=load_model('customer_behavior_model.keras')
predictions= new_model.predict(sc.transform(np.array([[42,11111000]])))[:,1]
predictions
from google.colab import files
files.download('customer_behavior_model.keras')
```
